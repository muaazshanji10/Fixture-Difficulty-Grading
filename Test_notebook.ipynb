```python
import os
from dotenv import load_dotenv
import requests

load_dotenv()

api_key = os.getenv("football_data_api_key")
```

```python
url = "https://api.football-data.org/v4/competitions/PL/standings"
headers = {"X-Auth-Token": api_key}
standings_response = requests.get(url, headers=headers)
print(standings_response.status_code)
print(standings_response.json())
```

```python
url = "https://api.football-data.org/v4/competitions/PL/matches"
headers = {"X-Auth-Token": api_key}
matches_response = requests.get(url, headers=headers)
print(matches_response.status_code)
print(matches_response.json())
```

```python
import json

os.makedirs("data/raw", exist_ok=True)
with open("data/raw/matches.json", "w") as f:
    json.dump(matches_response.json(), f, indent=2)

os.makedirs("data/raw", exist_ok=True)
with open("data/raw/standings.json", "w") as f:
    json.dump(standings_response.json(), f, indent=2)
```

```python
data_matches = matches_response.json()
print(data_matches.keys())                    # top-level structure
print(len(data_matches["matches"]))            # how many records
print(data_matches["matches"][0])              # just one example match, to see its fields
```

```python
data_standings = standings_response.json()
print(data_standings.keys())                    # top-level structure
print(len(data_standings["standings"]))            # how many records
print(data_standings["standings"][0])              # just one example match, to see its fields, realised it jus gives an entire table
```

```python
url = "https://api.football-data.org/v4/competitions/PL/matches?season=2022"
headers = {"X-Auth-Token": api_key}
response = requests.get(url, headers=headers)
print(response.status_code)
print(response.json())
```

```python
import duckdb

con = duckdb.connect() 
con.sql("""
    SELECT unnest(matches) AS match
    FROM read_json_auto('data/raw/matches.json')
    LIMIT 1
""").show()
```

```python
os.makedirs("data/raw", exist_ok=True)
with open("data/raw/match_1.json", "w") as f:
    json.dump(data_matches["matches"][0], f, indent=2)
```

```python
import duckdb

con = duckdb.connect()

con.sql("""
    CREATE TABLE matches_clean AS
    SELECT
        season.id as season_id,
        matchday as match_day,
        id as match_id,
        utcDate as date,
        homeTeam.name as home_team,
        awayTeam.name as away_team,
        Score.winner as winner
    FROM read_json_auto('data/raw/match_1.json')

""")

con.sql("SELECT * FROM matches_clean").show()
```

```python
os.makedirs("data/raw", exist_ok=True)
with open("data/raw/standings_present", "w") as f:
    json.dump(data_standings["standings"][0], f, indent=2)
```

```python
import duckdb

con = duckdb.connect()

con.sql("""
    CREATE TABLE standings_clean AS
    SELECT
        t.team.name AS team_name,
        t.position AS position,
        t.points AS points,
        t.goalDifference AS goal_difference
    FROM (
        SELECT unnest(standings[1].table) AS t
        FROM read_json_auto('data/raw/standings.json')
    )
""")

con.sql("SELECT * FROM standings_clean").show()
```
